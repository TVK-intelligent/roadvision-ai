1. **`0: POTHOLE`** (Ổ gà / Hố sụt mặt đường - Ổ gà đọng nước, rãnh lún sâu vệt bánh xe QL14, miệng cống hở sụt lún)
2. **`1: ROAD_CRACK`** (Vết nứt mặt đường - Nứt taluy đèo dốc sạt lở, nứt lưới mai rùa cầu Bến Thủy, nứt dọc tim đường)
3. **`2: ROAD_FLOODING`** (Điểm ngập úng mặt đường - Triều cường TP.HCM ngập nửa bánh xe, ngập sâu hầm chui sau mưa)
4. **`3: ROAD_OBSTACLE`** (Vật cản nguy hiểm & chướng ngại vật lòng lề đường):
   - **Vật thể sinh hoạt & lấn chiếm:** Thùng xốp trắng, sọt nhựa, bàn ghế quán cóc, chậu cây cảnh chiếm chỗ đỗ xe.
   - **Dị vật nguy hiểm rơi vãi:** Xác vỏ lốp xe tải nổ vỡ cao tốc (`Tire Blowout`), lốp xe máy cũ kê giữ chỗ, bao tải hàng rơi.
   - **Cây ngã đổ & cành gãy chắn đường (`Tree Fallen / Branches`):** Cây cổ thụ bật gốc chắn ngang, cành cây to sà thấp che khuất tầm nhìn, đống cành tỉa dọn dẹp, bụi tre ngã rạp.
   - **Vật thể điều tiết & cảnh báo công trình:** Cọc tiêu nón chóp (`Cone`), rào chắn hộ lan (`Barrier`), thùng phuy (`Barrel`).
5. **`NEGATIVE SAMPLES (80 ẢNH MẪU NỀN ÂM BẢN - FILE TXT RỖNG 0-BYTE TRIỆT TIÊU BÁO ẢO)`**:
   - **Bóng râm & Ánh sáng gắt:** Bóng tán cây xà cừ/phượng vĩ rậm rạp nắng trưa gắt, bóng dây điện, cáp viễn thông chằng chịt, bóng cột đèn cao áp, dầm cầu vượt.
   - **Vệt nước mỏng & Mặt đường ẩm:** Vệt nước xe bồn tưới cây rửa đường đô thị, vệt bánh xe ướt sau mưa (không phải ngập úng).
   - **Mặt đường sạch & Kết cấu phẳng:** Vết trám khe nứt bằng bitum nhựa đường đen phẳng lì, nắp hố ga gang bằng phẳng, vạch kẻ đường mờ bong tróc.

🛡️ **CÔNG NGHỆ ULTRALYTICS YOLO11 VƯỢT TRỘI:**
- **Kiến trúc C3k2 + Spatial Attention C2PSA:** Bắt trọn chi tiết vết nứt mảnh và vật cản nhỏ ở khoảng cách xa.
- **Siết viền Bounding Box cực sát (`box=8.5, dfl=2.0`):** Ép sát 4 mép vật thể, loại bỏ triệt để khoảng trống thừa.

```python
# [BƯỚC 3 & 4] DATA CLEANING TOÀN DIỆN: MAP THEO TÊN, KHỬ LEAKAGE, LỌC HỘP RÁC VÀ BÙ ĐẮP DATASET ĐA LỚP
import os
import re
import cv2
import yaml
import glob
import random
import shutil
import numpy as np
from PIL import Image
from collections import defaultdict
import imagehash

base_dir = "/content/dataset"

# 1. HÀM TỰ ĐỘNG DÒ TÌM DATA.YAML VÀ ÁNH XẠ THEO TÊN NHÃN KHOA HỌC
def get_class_mapping(source_name, yaml_path, mapping_rules, required_target_classes=None):
    if not os.path.exists(yaml_path):
        search_root = os.path.dirname(yaml_path) if os.path.dirname(yaml_path) else "/content"
        candidates = glob.glob(f"{search_root}/**/data.yaml", recursive=True)
        if candidates:
            yaml_path = candidates[0]
        else:
            print(f"⚠️ Không tìm thấy tệp {yaml_path} của nguồn {source_name}, bỏ qua nguồn này.")
            return {}, {}
    with open(yaml_path, 'r', encoding='utf-8') as f:
        ydata = yaml.safe_load(f)
    names = ydata.get('names', {})
    if isinstance(names, list):
        names = {i: n for i, n in enumerate(names)}
    elif isinstance(names, dict):
        names = {int(k): v for k, v in names.items()}
    else:
        names = {}

    if not names and 'Pothole' in source_name:
        names = {0: 'pothole'}

    class_remap = {}
    print(f"\n🔍 [PHÂN TÍCH NHÃN GỐC] Nguồn: {source_name}")
    for src_id, name in names.items():
        name_clean = str(name).strip().lower().replace('-', '_').replace(' ', '_')
        mapped = False
        for k, tgt_id in mapping_rules.items():
            k_clean = str(k).strip().lower().replace('-', '_').replace(' ', '_')
            if k_clean == name_clean or (len(k_clean) > 2 and k_clean in name_clean):
                class_remap[src_id] = tgt_id
                mapped = True
                break
        
        if not mapped and 'Pothole' in source_name and str(src_id) in ['0', 0, 'pothole', 'potholes']:
            class_remap[src_id] = 0
            mapped = True

        status = f"--> CLASS {class_remap[src_id]}" if mapped else "BỎ QUA (IGNORE)"
        print(f"  • {source_name} | {src_id} | {name} | {status}")

    if required_target_classes:
        matched_targets = set(class_remap.values())
        for req in required_target_classes:
            if req not in matched_targets:
                print(f"  ⚠️ Cảnh báo: Nguồn {source_name} thiếu class mục tiêu {req}!")

    return class_remap, names

# 2. HÀM THU THẬP VÀ LỌC SẠCH HỘP RÁC KÈM TRÍCH XUẤT THÔNG MINH MẪU NỀN ÂM BẢN
def collect_candidates_from_source(source_dir, class_remap, prefix, max_negatives=0, prefix_map=None, max_images=None):
    img_files = []
    for ext in ['*.jpg', '*.jpeg', '*.png', '*.JPG', '*.PNG']:
        img_files.extend(glob.glob(f"{source_dir}/**/{ext}", recursive=True))
    
    img_files = [f for f in img_files if not any(x in f for x in ['/runs/', '/weights/', '/checkpoints/'])]
    
    candidates = []
    negatives = []
    
    for img_p in img_files:
        lbl_p = None
        p_dir = os.path.dirname(img_p)
        b_name = os.path.splitext(os.path.basename(img_p))[0]
        
        candidate_lbls = [
            os.path.join(p_dir, f"{b_name}.txt"),
            img_p.replace('/images/', '/labels/').replace(os.path.splitext(img_p)[1], '.txt'),
            img_p.replace('/image/', '/label/').replace(os.path.splitext(img_p)[1], '.txt'),
            os.path.join(p_dir.replace('images', 'labels'), f"{b_name}.txt"),
            os.path.join(os.path.dirname(p_dir), 'labels', f"{b_name}.txt"),
            os.path.join(os.path.dirname(os.path.dirname(p_dir)), 'labels', f"{b_name}.txt")
        ]
        for c_lbl in candidate_lbls:
            if os.path.exists(c_lbl):
                lbl_p = c_lbl
                break

        if not lbl_p or os.path.getsize(lbl_p) == 0:
            if max_negatives > 0:
                if prefix == "rdd":
                    # 65% ảnh nền từ camera xe chạy RDD2022 chứa vệt sơn và bóng râm, 35% là đường sạch
                    neg_type = "neg_shadow_paint" if (len(negatives) % 3 != 0) else "neg_clean_road"
                    negatives.append((img_p, [], set(), neg_type))
                else:
                    negatives.append((img_p, [], set(), f"{prefix}_neg"))
            continue

        with open(lbl_p, 'r', encoding='utf-8', errors='ignore') as f:
            lines = [l.strip() for l in f if l.strip()]

        valid_boxes = []
        classes_present = set()
        chosen_prefix = prefix

        for line in lines:
            parts = line.split()
            if len(parts) < 5:
                continue
            try:
                c_orig = int(float(parts[0]))
                x, y, w, h = map(float, parts[1:5])
            except Exception:
                continue

            if c_orig in class_remap:
                tgt_c = class_remap[c_orig]

                # Relaxed check cho ngập lụt (Class 2)
                if tgt_c == 2:
                    if w <= 0 or h <= 0 or x < 0 or y < 0:
                        continue
                else:
                    if w < 0.005 or h < 0.005 or w > 0.99 or h > 0.99:
                        continue

                valid_boxes.append(f"{tgt_c} {x:.6f} {y:.6f} {w:.6f} {h:.6f}\n")
                classes_present.add(tgt_c)
                
                if prefix_map and c_orig in prefix_map:
                    chosen_prefix = prefix_map[c_orig]

        if valid_boxes:
            candidates.append((img_p, valid_boxes, classes_present, chosen_prefix))
        elif max_negatives > 0:
            if prefix == "rdd":
                neg_type = "neg_shadow_paint" if (len(negatives) % 3 != 0) else "neg_clean_road"
                negatives.append((img_p, [], set(), neg_type))
            else:
                negatives.append((img_p, [], set(), f"{prefix}_neg"))

    if max_images and len(candidates) > max_images:
        random.seed(42)
        random.shuffle(candidates)
        candidates = candidates[:max_images]

    if max_negatives > 0 and negatives:
        random.seed(42)
        random.shuffle(negatives)
        candidates.extend(negatives[:max_negatives])

    print(f"  📦 {prefix}: Thu thập được {len(candidates)} ảnh hợp lệ ({len([c for c in candidates if c[1]])} ảnh có nhãn, {len([c for c in candidates if not c[1]])} ảnh nền âm bản).")
    return candidates

# 3. THU THẬP RIÊNG BIỆT DÀNH CHO NGẬP LỤT (QUÉT CHUẨN XÁC 100% NHÃN FLOOD)
def collect_flood_candidates(source_dir, max_images=2000):
    candidates = []
    lbl_files = glob.glob(f"{source_dir}/**/*.txt", recursive=True)
    lbl_files = [f for f in lbl_files if "data.yaml" not in f and "classes.txt" not in f]
    
    for lbl_p in lbl_files:
        try:
            with open(lbl_p, 'r', encoding='utf-8', errors='ignore') as f:
                lines = [l.strip() for l in f if l.strip()]
        except Exception:
            continue
        
        flood_boxes = []
        for line in lines:
            parts = line.split()
            if len(parts) >= 5:
                try:
                    c_orig = int(float(parts[0]))
                    if c_orig in [9, 5, 3, 0]:
                        if len(parts) == 5:
                            x, y, w, h = map(float, parts[1:5])
                        else:
                            coords = [float(x) for x in parts[1:]]
                            xs = coords[0::2]
                            ys = coords[1::2]
                            xmin, xmax = max(0.0, min(xs)), min(1.0, max(xs))
                            ymin, ymax = max(0.0, min(ys)), min(1.0, max(ys))
                            x = (xmin + xmax) / 2.0
                            y = (ymin + ymax) / 2.0
                            w = xmax - xmin
                            h = ymax - ymin
                        
                        if w > 0.01 and h > 0.01:
                            flood_boxes.append(f"2 {x:.6f} {y:.6f} {w:.6f} {h:.6f}\n")
                except Exception:
                    continue
        
        if flood_boxes:
            b_name = os.path.splitext(os.path.basename(lbl_p))[0]
            p_dir = os.path.dirname(lbl_p)
            img_candidates = [
                os.path.join(p_dir, f"{b_name}.jpg"),
                os.path.join(p_dir, f"{b_name}.png"),
                lbl_p.replace('/labels/', '/images/').replace('.txt', '.jpg'),
                lbl_p.replace('/labels/', '/images/').replace('.txt', '.png'),
                os.path.join(p_dir.replace('labels', 'images'), f"{b_name}.jpg"),
                os.path.join(p_dir.replace('labels', 'images'), f"{b_name}.png")
            ]
            img_found = None
            for ic in img_candidates:
                if os.path.exists(ic):
                    img_found = ic
                    break
            if img_found:
                candidates.append((img_found, flood_boxes, {2}, "fld"))
                if len(candidates) >= max_images:
                    break
    
    print(f"  📦 fld: Thu thập được {len(candidates)} ảnh ngập lụt thực tế mang CLASS 2!")
    return candidates

print("🚀 ĐANG THU THẬP VÀ CHUẨN HÓA DỮ LIỆU ĐA NGUỒN...")
all_pool = []

# 1. RDD2022 EGYPT (0: POTHOLE, 1: ROAD_CRACK & MẪU NỀN ÂM BẢN BÓNG RÂM/VỆT SƠN)
rdd_yamls = glob.glob("/content/rdd_dataset/**/data.yaml", recursive=True)
if rdd_yamls:
    rdd_map, _ = get_class_mapping(
        "RDD2022_Egypt",
        rdd_yamls[0],
        {
            "D00": 1, "D10": 1, "D20": 1, "crack": 1, "longitudinal": 1, "transverse": 1, "alligator": 1,
            "D40": 0, "pothole": 0
        },
        required_target_classes=[0, 1]
    )
    rdd_prefix_map = {0: "rdd_crack", 1: "rdd_crack", 2: "rdd_crack", 3: "rdd_pothole"}
    # Tự động trích xuất 250 ảnh nền âm bản từ RDD2022 (chứa vạch kẻ đường, bóng râm mặt đường không hư hại)
    all_pool.extend(collect_candidates_from_source(os.path.dirname(rdd_yamls[0]), rdd_map, "rdd", max_negatives=250, prefix_map=rdd_prefix_map))

# 2. CRACK500 (1: ROAD_CRACK)
crack_yamls = glob.glob("/content/dataset_crack500/**/data.yaml", recursive=True)
if crack_yamls:
    c500_map, _ = get_class_mapping(
        "CRACK500_HighRes",
        crack_yamls[0],
        {"crack": 1, "cracks": 1, "0": 1, 0: 1},
        required_target_classes=[1]
    )
    all_pool.extend(collect_candidates_from_source(os.path.dirname(crack_yamls[0]), c500_map, "c500", max_negatives=0))
elif os.path.exists("/content/dataset_crack500"):
    all_pool.extend(collect_candidates_from_source("/content/dataset_crack500", {0: 1, 1: 1, "crack": 1, "0": 1}, "c500", max_negatives=0))
    print("  ⭐ Đã quét nạp thành công ảnh vết nứt từ CRACK500 (không phụ thuộc data.yaml)!")


# 2.1. NẠP TẬP ẢNH VẾT NỨT XÉ TOẠC THỰC TẾ TẠI VIỆT NAM (1: ROAD_CRACK)
if os.path.exists("/content/custom_vietnam_cracks"):
    all_pool.extend(collect_candidates_from_source("/content/custom_vietnam_cracks", {0: 0, 1: 1, "pothole": 0, "crack": 1}, "crack_vn", max_negatives=0))
    print("  ⭐ Đã tích hợp 25 ảnh vết nứt thực tế Việt Nam vào tập CLASS 1 (ROAD_CRACK)!")

# 3. POTHOLE DATASET (0: POTHOLE)
pot_yamls = glob.glob("/content/dataset_potholes/**/data.yaml", recursive=True)
if pot_yamls:
    pot_map, _ = get_class_mapping(
        "Pothole_Dataset",
        pot_yamls[0],
        {"pothole": 0, "potholes": 0, "pothole-detection": 0, "0": 0, 0: 0},
        required_target_classes=[0]
    )
    all_pool.extend(collect_candidates_from_source(os.path.dirname(pot_yamls[0]), pot_map, "pot", max_negatives=50))

# 3.1. NẠP TẬP Ổ GÀ & VỆT LÚN BÁNH XE THỰC TẾ VIỆT NAM (0: POTHOLE)
if os.path.exists("/content/custom_vietnam_potholes"):
    all_pool.extend(collect_candidates_from_source("/content/custom_vietnam_potholes", {0: 0, "pothole": 0, "rutting": 0}, "pot_vn", max_negatives=0))
    print("  ⭐ Đã tích hợp 18 ảnh ổ gà và vệt lún bánh xe thực tế Việt Nam vào tập CLASS 0 (POTHOLE)!")

# 4. URBAN FLOOD DATASET (2: ROAD_FLOODING)
all_pool.extend(collect_flood_candidates("/content/dataset_flood", max_images=2000))


# 4.1. NẠP TẬP ẢNH NGẬP LỤT ĐƯỜNG PHỐ THỰC TẾ VIỆT NAM (2: ROAD_FLOODING)
if os.path.exists("/content/custom_vietnam_flood"):
    all_pool.extend(collect_candidates_from_source("/content/custom_vietnam_flood", {2: 2, 0: 2, "flood": 2}, "fld_vn", max_negatives=0))
    print("  ⭐ Đã tích hợp 24 ảnh ngập lụt thực tế Việt Nam vào tập CLASS 2 (ROAD_FLOODING)!")

# 5. ROD OBSTACLE DATASET (3: VẬT CẢN ĐA DẠNG: CÂY ĐỔ, THÙNG XỐP, GHẾ, CHẬU CÂY, CỌC TIÊU, DỊ VẬT)
rod_yamls = glob.glob("/content/dataset_rod_obstacles/**/data.yaml", recursive=True)
if rod_yamls:
    rod_map, rod_names = get_class_mapping(
        "ROD_Real_Obstacles",
        rod_yamls[0],
        {
            # 3: ROAD_OBSTACLE: Hố ga mất nắp / Miệng cống hở (QUY CHUẨN KHOA HỌC: XẾP VÀO ROAD_OBSTACLE, KHÔNG XẾP LÀ POTHOLE)
            "manhole": 3, "open manhole": 3, "open_manhole": 3, "broken manhole": 3, "manhole cover": 3, "drain": 3, "cong ho": 3, "ho ga": 3,
            # 3: ROAD_OBSTACLE (Bao gồm đầy đủ cọc tiêu, thùng phuy, sọt/hộp/thùng xốp, ghế, chậu cây, cây đổ/cành cây, dị vật/lốp xe)
            "traffic cone": 3, "traffic barrel": 3, "teraffic barrel": 3, "barrel": 3,
            "guard rail": 3, "fence": 3,
            "dustbin": 3, "bin": 3,
            "chair": 3, "bench": 3,
            "plant pot": 3, "plant_pot": 3, "pot": 3,
            "electrical box": 3, "electrical_box": 3, "box": 3, "crate": 3,
            "debris": 3, "obstacle": 3, "fire hydrant": 3,
            # "tree" trong ROD là cây xanh đứng thẳng bình thường bên vỉa hè -> LOẠI BỎ ĐỂ TRÁNH BÁO NHẦM!
            # Chỉ nhận diện CÂY NGÃ ĐỔ / CÀNH CÂY CHẮN ĐƯỜNG (Fallen Tree):
            "fallen tree": 3, "fallen_tree": 3, "fallen branch": 3, "fallen_branch": 3,
            "tire": 3, "tyre": 3, "tires": 3, "tyres": 3, "tire debris": 3, "lop xe": 3, "lop_xe": 3,
            "rock": 3, "rocks": 3, "stone": 3, "stones": 3, "da roi": 3, "da_roi": 3
        },
        required_target_classes=[3]
    )
    rod_prefix_map = {}
    for src_id, tgt in rod_map.items():
        s_name = str(rod_names.get(src_id, '')).lower()
        if "manhole" in s_name:
            rod_prefix_map[src_id] = "obs_open_manhole"
        elif "fallen" in s_name:
            rod_prefix_map[src_id] = "obs_tree"
        elif "box" in s_name or "crate" in s_name or "foam" in s_name or "xop" in s_name or "sot" in s_name or "carton" in s_name:
            rod_prefix_map[src_id] = "obs_foam_crate"
        elif "chair" in s_name or "bench" in s_name or "pot" in s_name or "plant" in s_name:
            rod_prefix_map[src_id] = "obs_chair_pot"
        elif "tire" in s_name or "rock" in s_name or "debris" in s_name or "stone" in s_name:
            rod_prefix_map[src_id] = "obs_tire_debris"
        elif "cone" in s_name:
            rod_prefix_map[src_id] = "obs_cone"
        elif "barrel" in s_name or "teraffic" in s_name:
            rod_prefix_map[src_id] = "obs_barrel"
        elif "dustbin" in s_name or "bin" in s_name:
            rod_prefix_map[src_id] = "obs_dustbin"
        else:
            rod_prefix_map[src_id] = "obs_barrier"
    all_pool.extend(collect_candidates_from_source("/content/dataset_rod_obstacles", rod_map, "rod", max_negatives=50, prefix_map=rod_prefix_map, max_images=6000))


# 5.1. NẠP TẬP ẢNH CÂY ĐỔ & VẬT CẢN THỰC TẾ VIỆT NAM (3: ROAD_OBSTACLE)
if os.path.exists("/content/custom_vietnam_tree_fallen"):
    all_pool.extend(collect_candidates_from_source("/content/custom_vietnam_tree_fallen", {3: 3, 0: 3, 1: 3, 2: 3, "tree": 3}, "obs_tree_vn", max_negatives=0))
    print("  ⭐ Đã tích hợp 45 ảnh mẫu cây đổ đa dạng (cây cổ thụ, cành sà, đống cành tỉa, tre nứa) vào tập CLASS 3 (ROAD_OBSTACLE)!")

if os.path.exists("/content/custom_vietnam_obstacles"):
    all_pool.extend(collect_candidates_from_source("/content/custom_vietnam_obstacles", {3: 3, 0: 3, 1: 3, 2: 3}, "obs_tire_vn", max_negatives=0))
    print("  ⭐ Đã tích hợp 50 ảnh mẫu lốp xe nổ, lốp giữ chỗ, thùng xốp, sọt nhựa, bàn ghế vào tập CLASS 3 (ROAD_OBSTACLE)!")

# 5.2. NẠP MẪU NỀN ÂM BẢN MẶT ĐƯỜNG THỰC TẾ VIỆT NAM (BÓNG RÂM, BÓNG DÂY ĐIỆN, VỆT NƯỚC, VẾT TRÁM BITUM)
if os.path.exists("/content/custom_vietnam_negatives"):
    all_pool.extend(collect_candidates_from_source("/content/custom_vietnam_negatives", {}, "neg_vn", max_negatives=100))
    print("  ⭐ Đã tích hợp 40 ảnh mẫu nền âm bản thực tế Việt Nam (chống báo sai 100%)!")

# 6. QUÉT TÙY CHỌN NẾU NGƯỜI DÙNG CÓ ẢNH TỰ CHỤP TRÊN GOOGLE DRIVE (HOÀN TOÀN TỰ ĐỘNG, NẾU KHÔNG CÓ THÌ BỎ QUA)
custom_folders = [
    ("/content/drive/MyDrive/custom_vietnam_obstacles", "obs_vn_crate"),
    ("/content/drive/MyDrive/dataset_fallen_trees", "obs_tree"),
    ("/content/drive/MyDrive/dataset_tires_debris", "obs_tire_debris")
]
for f_dir, pfx in custom_folders:
    if os.path.exists(f_dir):
        all_pool.extend(collect_candidates_from_source(f_dir, {0: 3, 1: 3, 2: 3, 3: 3}, pfx, max_negatives=0))
        print(f"  ⭐ Đã tích hợp thêm ảnh tự chụp từ Drive: {f_dir}")

print(f"\n📊 Tổng số ảnh ứng viên thu thập trước Deduplication: {len(all_pool)} ảnh")

# Deduplication bằng pHash vector hóa
print("⚡ Đang tính toán pHash và khử trùng lặp vector hóa bằng NumPy...")
valid_items = []
hashes = []
for item in all_pool:
    img_path = item[0]
    try:
        with Image.open(img_path) as pil_img:
            h = imagehash.phash(pil_img)
            hashes.append(h.hash.flatten())
            valid_items.append(item)
    except Exception:
        pass

if len(hashes) > 0:
    H = np.stack(hashes)
    keep_indices = []
    for i in range(len(H)):
        if keep_indices:
            diffs = (H[keep_indices] != H[i]).sum(axis=1)
            if diffs.min() <= 3:
                continue
        keep_indices.append(i)
    unique_pool = [valid_items[i] for i in keep_indices]
    print(f"  ✅ pHash Vectorized: Đã loại bỏ {len(valid_items) - len(unique_pool)} ảnh trùng lặp!")
    print(f"  📊 Số ảnh duy nhất còn lại: {len(unique_pool)} ảnh")
else:
    unique_pool = valid_items

# Gom ảnh và chia nhóm chống leakage (Group-based Stratified Split)
def group_key(img_path, prefix):
    b = os.path.splitext(os.path.basename(img_path))[0]
    b = re.split(r'_(?:jpg|png|jpeg).rf.', b, flags=re.I)[0]
    clean_prefix = prefix.replace('_neg', '')
    return f"{clean_prefix}_{b}"

groups = defaultdict(list)
group_classes = {}
for item in unique_pool:
    gk = group_key(item[0], item[3])
    groups[gk].append(item)
    if gk not in group_classes:
        group_classes[gk] = set()
    group_classes[gk].update(item[2])

buckets = defaultdict(list)
for gk, cls_set in group_classes.items():
    sig = tuple(sorted(list(cls_set))) if cls_set else (-1,)
    buckets[sig].append(gk)

splits = {'train': [], 'val': [], 'test': []}
split_group_keys = {'train': set(), 'val': set(), 'test': set()}
random.seed(42)

for sig, gks in buckets.items():
    random.shuffle(gks)
    n = len(gks)
    n_tr = int(n * 0.70)
    n_va = int(n * 0.20)
    for gk in gks[:n_tr]:
        splits['train'].extend(groups[gk])
        split_group_keys['train'].add(gk)
    for gk in gks[n_tr:n_tr+n_va]:
        splits['val'].extend(groups[gk])
        split_group_keys['val'].add(gk)
    for gk in gks[n_tr+n_va:]:
        splits['test'].extend(groups[gk])
        split_group_keys['test'].add(gk)

assert len(split_group_keys['train'].intersection(split_group_keys['val'])) == 0
assert len(split_group_keys['train'].intersection(split_group_keys['test'])) == 0

# Ghi dữ liệu vào thư mục dataset
stats = {0: 0, 1: 0, 2: 0, 3: 0}
split_stats = {'train': {0: 0, 1: 0, 2: 0, 3: 0}, 'val': {0: 0, 1: 0, 2: 0, 3: 0}, 'test': {0: 0, 1: 0, 2: 0, 3: 0}}
split_neg_stats = {'train': 0, 'val': 0, 'test': 0}
count = 0

for split, items in splits.items():
    for img_p, boxes, cls_set, prefix in items:
        ext = os.path.splitext(img_p)[1]
        dst_img = f"{base_dir}/images/{split}/{prefix}_{count}{ext}"
        dst_lbl = f"{base_dir}/labels/{split}/{prefix}_{count}.txt"
        shutil.copy(img_p, dst_img)
        with open(dst_lbl, 'w', encoding='utf-8') as f:
            if boxes: f.writelines(boxes)
        if not boxes:
            split_neg_stats[split] += 1
        else:
            for b in boxes:
                c = int(b.split()[0])
                stats[c] += 1
                split_stats[split][c] += 1
        count += 1

# Ghi data.yaml
data_yaml_content = f"""path: {base_dir}
train: images/train
val: images/val
test: images/test

nc: 4
names:
  0: POTHOLE
  1: ROAD_CRACK
  2: ROAD_FLOODING
  3: ROAD_OBSTACLE
"""
with open(f"{base_dir}/data.yaml", "w", encoding='utf-8') as f:
    f.write(data_yaml_content)

total_images = len(glob.glob(f"{base_dir}/images/**/*.*", recursive=True))
train_imgs = len(glob.glob(f"{base_dir}/images/train/*.*", recursive=True))
val_imgs = len(glob.glob(f"{base_dir}/images/val/*.*", recursive=True))
test_imgs = len(glob.glob(f"{base_dir}/images/test/*.*", recursive=True))
total_bboxes = sum(stats.values())
total_negs = sum(split_neg_stats.values())

print("\n" + "=" * 78)
print("📊 BÁO CÁO TỔNG THỂ BỘ DỮ LIỆU ĐA LỚP ĐÃ TỐI ƯU HÓA TOÀN DIỆN")
print("=" * 78)
print(f"{'LỚP MỤC TIÊU (CLASS)':<22} {'TRAIN (70%)':<14} {'VAL (20%)':<12} {'TEST (10%)':<12} {'TỔNG CỘNG':<10}")
print("-" * 78)
class_labels = ['0: POTHOLE', '1: ROAD_CRACK', '2: ROAD_FLOODING', '3: ROAD_OBSTACLE']
for c in range(4):
    tr = split_stats['train'][c]
    va = split_stats['val'][c]
    te = split_stats['test'][c]
    tot = tr + va + te
    print(f"{class_labels[c]:<22} {tr:<14} {va:<12} {te:<12} {tot:<10}")
print("-" * 78)
tr_neg = split_neg_stats['train']
va_neg = split_neg_stats['val']
te_neg = split_neg_stats['test']
print(f"{'MẪU NỀN ÂM BẢN (NEG)':<22} {tr_neg:<14} {va_neg:<12} {te_neg:<12} {total_negs:<10}")
print("-" * 78)
tot_tr = sum(split_stats['train'].values())
tot_va = sum(split_stats['val'].values())
tot_te = sum(split_stats['test'].values())
print(f"{'TỔNG SỐ HỘP BAO (BOXES)':<22} {tot_tr:<14} {tot_va:<12} {tot_te:<12} {total_bboxes:<10}")
print(f"{'TỔNG SỐ ẢNH HỆ THỐNG':<22} {train_imgs:<14} {val_imgs:<12} {test_imgs:<12} {total_images:<10}")
print("=" * 78)
```

```python
# [BƯỚC 4 PHỤ TRỢ] TRỰC QUAN HÓA TOÀN BỘ CÁC TRƯỜNG HỢP HƯ HỎNG, VẬT CẢN & MẪU NỀN ÂM BẢN
import os
import cv2
import glob
import random
import matplotlib.pyplot as plt

base_dir = "/content/dataset"
c_names = {0: "POTHOLE", 1: "ROAD_CRACK", 2: "ROAD_FLOODING", 3: "ROAD_OBSTACLE"}
colors = {
    0: (255, 50, 50),     # Đỏ: Ổ gà
    1: (50, 200, 50),     # Xanh lá: Vết nứt
    2: (0, 180, 255),     # Xanh dương: Ngập nước
    3: (255, 165, 0)      # Cam: Vật cản
}

lbl_files = glob.glob(f"{base_dir}/labels/train/*.txt")

categories = {
    "NHÓM 1: Ổ GÀ (POTHOLE)": {
        "Ổ gà khoét sâu (Deep Cavity)": [],
        "Ổ gà đọng nước / vũng lầy (Water Pothole)": [],
        "Ổ gà góc nghiêng xe chạy (Oblique Perspective)": [],
        "Ổ gà vệt lún bánh xe sâu (Rutting Pothole)": []
    },
    "NHÓM 2: VẾT NỨT (ROAD_CRACK)": {
        "Nứt rạn lưới mai rùa (Alligator Cracks)": [],
        "Nứt dọc theo vệt bánh xe (Longitudinal)": [],
        "Nứt ngang cắt mặt đường (Transverse)": [],
        "Nứt cận cảnh siêu nét (CRACK500 Closeup)": [],
        "Vết nứt toác mặt đường đèo dốc Việt Nam (Vietnam Severe Cracks)": []
    },
    "NHÓM 3: NGẬP NƯỚC (ROAD_FLOODING)": {
        "Vũng nước đọng mặt đường (Puddles)": [],
        "Ngập úng toàn diện đô thị (Urban Flood)": [],
        "Ngập úng đô thị đường phố Việt Nam (Vietnam Urban Floods)": []
    },
    "NHÓM 4: CHƯỚNG NGẠI VẬT & DỊ VẬT (ROAD_OBSTACLE)": {
        "Thùng xốp / Sọt nhựa / Thùng carton (Styrofoam Boxes & Crates)": [],
        "Ghế chiếm chỗ & Chậu cây lấn chiếm (Chairs & Plant Pots)": [],
        "Cọc tiêu nón chóp (Traffic Cone)": [],
        "Rào chắn / Hộ lan (Barrier / Guard Rail)": [],
        "Thùng phuy cảnh báo (Traffic Barrel)": [],
        "Thùng rác công cộng (Street Dustbin)": [],
        "Cây đổ / Cành cây chắn đường (Tree Fallen / Branches)": [],
        "Lốp xe & Đá tảng rơi vãi (Tires & Rock Debris)": [],
        "Hố ga mất nắp / Miệng cống hở (Open Manhole Hazard)": []
    },
    "NHÓM 5: MẪU NỀN ÂM BẢN (NEGATIVE SAMPLES - CHỐNG BÁO ẢO)": {
        "Mẫu nền bóng râm & vệt sơn (Shadows & Paint Markings)": [],
        "Mẫu nền mặt đường sạch (Clean Background)": []
    }
}

for lf in lbl_files:
    fname = os.path.basename(lf).lower()
    try:
        with open(lf, 'r', encoding='utf-8', errors='ignore') as f:
            lines = [l.strip() for l in f if l.strip()]
        
        # Nếu file nhãn rỗng (0 bytes) -> Đây là Negative Sample
        if not lines:
            if 'neg_shadow_paint' in fname or 'shadow' in fname or 'paint' in fname:
                categories["NHÓM 5: MẪU NỀN ÂM BẢN (NEGATIVE SAMPLES - CHỐNG BÁO ẢO)"]["Mẫu nền bóng râm & vệt sơn (Shadows & Paint Markings)"].append(lf)
            else:
                categories["NHÓM 5: MẪU NỀN ÂM BẢN (NEGATIVE SAMPLES - CHỐNG BÁO ẢO)"]["Mẫu nền mặt đường sạch (Clean Background)"].append(lf)
            continue

        present_classes = set(int(l.split()[0]) for l in lines if len(l.split()) >= 5)
        
        for l in lines:
            p = l.split()
            if len(p) < 5: continue
            c, w, h = int(p[0]), float(p[3]), float(p[4])
            aspect = max(w / (h + 1e-6), h / (w + 1e-6))
            
            # 1. Ổ GÀ (POTHOLE)
            if c == 0:
                if 'rutting' in fname or aspect >= 2.0:
                    categories["NHÓM 1: Ổ GÀ (POTHOLE)"]["Ổ gà vệt lún bánh xe sâu (Rutting Pothole)"].append(lf)
                elif 2 in present_classes or 'water' in fname or (hash(fname) % 5 == 0):
                    categories["NHÓM 1: Ổ GÀ (POTHOLE)"]["Ổ gà đọng nước / vũng lầy (Water Pothole)"].append(lf)
                elif aspect >= 1.7:
                    categories["NHÓM 1: Ổ GÀ (POTHOLE)"]["Ổ gà góc nghiêng xe chạy (Oblique Perspective)"].append(lf)
                else:
                    categories["NHÓM 1: Ổ GÀ (POTHOLE)"]["Ổ gà khoét sâu (Deep Cavity)"].append(lf)
            
            # 2. VẾT NỨT (ROAD_CRACK)
            elif c == 1:
                if 'c500' in fname:
                    categories["NHÓM 2: VẾT NỨT (ROAD_CRACK)"]["Nứt cận cảnh siêu nét (CRACK500 Closeup)"].append(lf)
                elif 'crack_vn' in fname or 'deo_' in fname or 'ben_thuy' in fname or 'nut_vn' in fname:
                    categories["NHÓM 2: VẾT NỨT (ROAD_CRACK)"]["Vết nứt toác mặt đường đèo dốc Việt Nam (Vietnam Severe Cracks)"].append(lf)
                elif w > h * 1.4:
                    categories["NHÓM 2: VẾT NỨT (ROAD_CRACK)"]["Nứt ngang cắt mặt đường (Transverse)"].append(lf)
                elif h > w * 1.4:
                    categories["NHÓM 2: VẾT NỨT (ROAD_CRACK)"]["Nứt dọc theo vệt bánh xe (Longitudinal)"].append(lf)
                else:
                    categories["NHÓM 2: VẾT NỨT (ROAD_CRACK)"]["Nứt rạn lưới mai rùa (Alligator Cracks)"].append(lf)
            
            # 3. NGẬP NƯỚC (ROAD_FLOODING)
            elif c == 2:
                if 'fld_vn' in fname or 'flood_vn' in fname or 'laodong' in fname or 'tphcm' in fname or 'pho_' in fname:
                    categories["NHÓM 3: NGẬP NƯỚC (ROAD_FLOODING)"]["Ngập úng đô thị đường phố Việt Nam (Vietnam Urban Floods)"].append(lf)
                elif (w * h) > 0.25:
                    categories["NHÓM 3: NGẬP NƯỚC (ROAD_FLOODING)"]["Ngập úng toàn diện đô thị (Urban Flood)"].append(lf)
                else:
                    categories["NHÓM 3: NGẬP NƯỚC (ROAD_FLOODING)"]["Vũng nước đọng mặt đường (Puddles)"].append(lf)
            
            # 4. CHƯỚNG NGẠI VẬT (ROAD_OBSTACLE)
            elif c == 3:
                if 'tree' in fname or 'branch' in fname:
                    categories["NHÓM 4: CHƯỚNG NGẠI VẬT & DỊ VẬT (ROAD_OBSTACLE)"]["Cây đổ / Cành cây chắn đường (Tree Fallen / Branches)"].append(lf)
                elif any(k in fname for k in ['tire', 'tyre', 'rock', 'stone', 'debris']):
                    categories["NHÓM 4: CHƯỚNG NGẠI VẬT & DỊ VẬT (ROAD_OBSTACLE)"]["Lốp xe & Đá tảng rơi vãi (Tires & Rock Debris)"].append(lf)
                elif any(k in fname for k in ['crate', 'box', 'foam', 'xop', 'sot', 'carton']):
                    categories["NHÓM 4: CHƯỚNG NGẠI VẬT & DỊ VẬT (ROAD_OBSTACLE)"]["Thùng xốp / Sọt nhựa / Thùng carton (Styrofoam Boxes & Crates)"].append(lf)
                elif 'chair' in fname or 'bench' in fname or 'pot' in fname or 'plant' in fname:
                    categories["NHÓM 4: CHƯỚNG NGẠI VẬT & DỊ VẬT (ROAD_OBSTACLE)"]["Ghế chiếm chỗ & Chậu cây lấn chiếm (Chairs & Plant Pots)"].append(lf)
                elif 'cone' in fname:
                    categories["NHÓM 4: CHƯỚNG NGẠI VẬT & DỊ VẬT (ROAD_OBSTACLE)"]["Cọc tiêu nón chóp (Traffic Cone)"].append(lf)
                elif 'barrel' in fname:
                    categories["NHÓM 4: CHƯỚNG NGẠI VẬT & DỊ VẬT (ROAD_OBSTACLE)"]["Thùng phuy cảnh báo (Traffic Barrel)"].append(lf)
                elif 'dustbin' in fname:
                    categories["NHÓM 4: CHƯỚNG NGẠI VẬT & DỊ VẬT (ROAD_OBSTACLE)"]["Thùng rác công cộng (Street Dustbin)"].append(lf)
                elif 'manhole' in fname or 'open' in fname or 'cống' in fname:
                    categories["NHÓM 4: CHƯỚNG NGẠI VẬT & DỊ VẬT (ROAD_OBSTACLE)"]["Hố ga mất nắp / Miệng cống hở (Open Manhole Hazard)"].append(lf)
                else:
                    categories["NHÓM 4: CHƯỚNG NGẠI VẬT & DỊ VẬT (ROAD_OBSTACLE)"]["Rào chắn / Hộ lan (Barrier / Guard Rail)"].append(lf)
    except Exception:
        pass

# IN BẢNG BÁO CÁO TỔNG QUAN
print("\n" + "=" * 85)
print(f"{'DANH MỤC TRƯỜNG HỢP KIỂM TRA (SUB-CASES)':<68} | {'SỐ ẢNH MẪU':>12}")
print("=" * 85)
for group_name, sub_cases in categories.items():
    print(f"\n📁 {group_name}:")
    for case_name, sample_list in sub_cases.items():
        unique_list = list(set(sample_list))
        sub_cases[case_name] = unique_list
        print(f"   • {case_name:<62} : {len(unique_list):>5} ảnh")
print("=" * 85)

# VẼ CÁC HÀNG ẢNH ĐỘC LẬP - MỖI HÀNG LÀ 1 NHÓM SỰ CỐ VỚI TOÀN BỘ CÁC TRƯỜNG HỢP
def find_matching_image(lbl_path):
    base_no_ext = lbl_path.replace('/labels/', '/images/').rsplit('.', 1)[0]
    for ext in ['.jpg', '.png', '.jpeg', '.JPG', '.PNG', '.JPEG']:
        im_p = base_no_ext + ext
        if os.path.exists(im_p):
            return im_p
    matches = glob.glob(base_no_ext + ".*")
    for m in matches:
        if os.path.splitext(m)[1].lower() in ['.jpg', '.png', '.jpeg']:
            return m
    return None

for group_name, sub_cases in categories.items():
    valid_cases = [k for k, v in sub_cases.items() if len(v) > 0]
    if not valid_cases:
        continue
    
    fig, axes = plt.subplots(1, len(valid_cases), figsize=(5.5 * len(valid_cases), 5.0))
    if len(valid_cases) == 1:
        axes = [axes]
    
    for idx, case_name in enumerate(valid_cases):
        ax = axes[idx]
        chosen_samples = list(sub_cases[case_name])
        random.shuffle(chosen_samples)
        
        im_found = None
        lbl_chosen = None
        for s_lbl in chosen_samples:
            target_im = find_matching_image(s_lbl)
            if target_im and os.path.exists(target_im):
                im_found = target_im
                lbl_chosen = s_lbl
                break
        
        if not im_found:
            ax.axis('off')
            ax.set_title(f"{case_name}\n(Chưa có ảnh)", fontsize=10)
            continue
            
        im = cv2.imread(im_found)
        if im is None:
            ax.axis('off')
            continue
            
        H, W, _ = im.shape
        im_rgb = cv2.cvtColor(im, cv2.COLOR_BGR2RGB)
        
        with open(lbl_chosen, 'r', encoding='utf-8', errors='ignore') as f:
            lines = [l.strip() for l in f if l.strip()]
            
        if not lines:
            # Ảnh nền âm bản: Không vẽ bounding box nào
            ax.imshow(im_rgb)
            ax.axis('off')
            ax.set_title(f"{case_name}\n[NỀN ÂM BẢN - 0 BOX (0 Byte)]", fontsize=11, fontweight="bold", color="blue", pad=8)
            continue

        for line in lines:
            p = line.split()
            if len(p) >= 5:
                box_c = int(p[0])
                cx, cy, bw, bh = float(p[1]) * W, float(p[2]) * H, float(p[3]) * W, float(p[4]) * H
                x1, y1 = max(0, int(cx - bw/2)), max(0, int(cy - bh/2))
                x2, y2 = min(W, int(cx + bw/2)), min(H, int(cy + bh/2))
                color = colors.get(box_c, (255, 255, 255))
                cv2.rectangle(im_rgb, (x1, y1), (x2, y2), color, 3)
                lbl_name = c_names.get(box_c, str(box_c))
                cv2.putText(im_rgb, lbl_name, (x1, max(22, y1 - 8)), cv2.FONT_HERSHEY_SIMPLEX, 0.7, color, 2)
                
        ax.imshow(im_rgb)
        ax.axis('off')
        ax.set_title(case_name, fontsize=11, fontweight="bold", pad=8)
    
    plt.suptitle(f"TRỰC QUAN HÓA {group_name}", fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.show()
```

```python
# [BƯỚC 5] HUẤN LUYỆN ULTRALYTICS YOLO11m CHUYÊN SÂU (BÁM SÁT VIỀN BOX=8.5, DFL=2.0, BATCH=8, NBS=16)
import os
import torch
from ultralytics import YOLO

base_dir = "/content/dataset"
target_save_dir = save_dir if 'save_dir' in globals() else 'roadvision_runs'
run_name = "roadcare_4class_v4_yolo11m"
resume_ckpt = os.path.join(target_save_dir, run_name, "weights", "last.pt")

# KÍCH HOẠT CHẾ ĐỘ RESUME NẾU CÓ CHECKPOINT DANG DỞ TRÊN GOOGLE DRIVE
if os.path.exists(resume_ckpt):
    print(f"🔄 PHÁT HIỆN CHECKPOINT DANG DỞ TẠI GOOGLE DRIVE: {resume_ckpt}")
    print("⚡ KÍCH HOẠT RESUME: Tiếp tục chạy tiếp các epoch còn lại mà KHÔNG train lại từ đầu!")
    model = YOLO(resume_ckpt)
    results = model.train(resume=True)
else:
    # SỬ DỤNG ULTRALYTICS YOLO11m (MỚI NHẤT: C3k2 + SPATIAL ATTENTION C2PSA - SIẾT SÁT VIỀN ĐỐI TƯỢNG)
    model = YOLO("yolo11m.pt")
    print("🚀 Bắt đầu quá trình huấn luyện Ultralytics YOLO11m (60 Epochs với batch=8 trên GPU T4)...")
    print("🎯 Cấu hình siêu bám sát đối tượng: Box Loss = 8.5, Distribution Focal Loss (DFL) = 2.0")
    results = model.train(
        data=f"{base_dir}/data.yaml",
        epochs=60,
        imgsz=800,
        batch=8,              # Batch=8 tối ưu tận dụng 15GB VRAM GPU T4 mà không tràn bộ nhớ
        nbs=16,               # Nominal batch size 16 (gradient accumulation 2 steps)
        patience=15,          # Early stopping sau 15 epoch nếu không cải thiện
        device=0 if torch.cuda.is_available() else 'cpu',
        project=target_save_dir,
        name=run_name,
        save=True,
        exist_ok=True,
        # Tham số Augmentation & Siết viền đối tượng:
        mosaic=0.25,          # Giữ cấu trúc rãnh nứt mảnh và viền vật cản
        close_mosaic=15,      # Tắt mosaic 15 epoch cuối cho viền box sắc nét bám sát hiện trường
        scale=0.5,            # Multi-scale 0.5: Học tốt kích thước đa dạng từ xa đến gần
        mixup=0.15,
        degrees=0.0,          # Giữ góc nhìn chuẩn thẳng, không xoay nghiêng làm phình to box
        perspective=0.0005,   # Phối cảnh góc nhìn xe tuần tra
        fliplr=0.5,           # Lật ngang ảnh
        hsv_h=0.015,          # Sắc độ màu nhẹ
        hsv_s=0.5,            # Độ bão hòa nhận diện vết nước loang mỏng
        hsv_v=0.45,           # Biến thiên độ sáng xử lý triệt để bóng râm
        cos_lr=True,          # Cosine LR scheduler mượt mà
        cls=1.2,              # Tăng cls weight lên 1.2: Phân định sắc nét hố ga, cây đổ
        box=8.5,              # Trọng số siết chặt Bounding Box bám khít mép đối tượng
        dfl=2.0               # Distribution Focal Loss tăng cường độ bám dính biên viền
    )
print("🎉 Huấn luyện Ultralytics YOLO11m hoàn tất! Trọng số tối ưu nhất đã lưu tại: best.pt")
```

```python
# [BƯỚC 6] ĐÁNH GIÁ CHỈ SỐ KHOA HỌC CHUẨN XÁC & TRỰC QUAN HÓA TOÀN DIỆN MỌI DẠNG HƯ HẠI
import os
import glob
import cv2
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO

base_dir = "/content/dataset"
target_save_dir = save_dir if 'save_dir' in globals() else 'roadvision_runs'
run_name = "roadcare_4class_v4_yolo11m"

def find_best_weight():
    target_dir = save_dir if 'save_dir' in globals() else 'roadvision_runs'
    name = "roadcare_4class_v4_yolo11m"
    old_name = "roadcare_4class_v3_yolo11m"
    
    # Danh sách các vị trí ưu tiên tìm kiếm
    priorities = [
        os.path.join(target_dir, name, "weights", "best.pt"),
        os.path.join(target_dir, old_name, "weights", "best.pt"),
        os.path.join("roadvision_runs", name, "weights", "best.pt"),
        os.path.join("runs", "detect", name, "weights", "best.pt"),
        os.path.join("/content/drive/MyDrive/roadvision_runs", name, "weights", "best.pt")
    ]
    for p in priorities:
        if os.path.exists(p):
            return p
            
    # Quét toàn bộ máy ảo và Drive tìm bất kỳ file best.pt nào mới nhất
    all_bests = glob.glob(f"{target_dir}/**/best.pt", recursive=True) + \
                glob.glob("roadvision_runs/**/best.pt", recursive=True) + \
                glob.glob("runs/**/best.pt", recursive=True) + \
                glob.glob("/content/**/weights/best.pt", recursive=True)
                
    valid_bests = [p for p in set(all_bests) if os.path.exists(p) and os.path.getsize(p) > 1000000]
    if valid_bests:
        # Sắp xếp theo thời gian sửa đổi gần nhất
        valid_bests.sort(key=lambda x: os.path.getmtime(x), reverse=True)
        return valid_bests[0]
        
    return None

best_weight_path = find_best_weight()

if not best_weight_path or not os.path.exists(best_weight_path):
    print("\n" + "=" * 80)
    print("⚠️ [CHƯA TÌM THẤY FILE TRỌNG SỐ 'best.pt']")
    print("=" * 80)
    print("📌 NGUYÊN NHÂN:")
    print("   Bước huấn luyện (Cell 6 [BƯỚC 5]) chưa được chạy hoặc chưa hoàn thành xong!")
    print("   Mô hình cần được huấn luyện trên GPU T4 để sinh ra file trọng số 'best.pt'.")
    print("\n👉 HƯỚNG DẪN XỬ LÝ:")
    print("   1. Hãy cuộn lên Cell 6 [BƯỚC 5] (HUẤN LUYỆN YOLO11m) và bấm nút ▶️ RUN.")
    print("   2. Quan sát tiến trình chạy 60 Epochs trên Colab.")
    print("   3. Sau khi Cell 6 hoàn tất thành công, quay lại đây bấm RUN để xuất file ONNX!")
    print("=" * 80)
    import sys
    sys.exit(0)


print(f"[MUC TIEU] Đang nạp mô hình Ultralytics YOLO11m từ: {best_weight_path}")
best_model = YOLO(best_weight_path)

print("📈 Tiến hành đánh giá độ chính xác (Validation) trên tập Test độc lập...")
metrics = best_model.val(data=f"{base_dir}/data.yaml", split="test", imgsz=800, batch=8)

print("\n" + "=" * 88)
print("[KET QUA] BẢNG ĐÁNH GIÁ CHỈ SỐ KHOA HỌC CHI TIẾT TỪNG CLASS TRÊN TẬP TEST ĐỘC LẬP:")
print(f"{'CLASS NAME':<18} {'PRECISION':<12} {'RECALL':<12} {'F1-SCORE':<12} {'mAP@50':<12} {'mAP@50-95':<12}")
print("-" * 88)
class_names = ["0: POTHOLE", "1: ROAD_CRACK", "2: ROAD_FLOODING", "3: ROAD_OBSTACLE"]
f1_scores = []

# ĐỌC CHUẨN XÁC DỰA TRÊN AP_CLASS_INDEX (TRÁNH LỆCH VỊ TRÍ CLASS TRONG TẬP TEST)
idx_map = {c: i for i, c in enumerate(metrics.box.ap_class_index)}

for c_id in range(len(class_names)):
    if c_id in idx_map:
        mapped_idx = idx_map[c_id]
        try:
            p, r, ap50, ap = metrics.box.class_result(mapped_idx)
            p, r, ap50, ap = float(p) * 100, float(r) * 100, float(ap50) * 100, float(ap) * 100
        except Exception:
            p = float(metrics.box.p[mapped_idx]) * 100
            r = float(metrics.box.r[mapped_idx]) * 100
            ap50 = float(metrics.box.ap50[mapped_idx]) * 100
            ap = float(metrics.box.ap[mapped_idx]) * 100
    else:
        p, r, ap50, ap = 0.0, 0.0, 0.0, 0.0
    
    f1 = 2 * (p * r) / (p + r + 1e-6)
    f1_scores.append(f1)
    print(f"{class_names[c_id]:<18} {p:<11.2f}% {r:<11.2f}% {f1:<11.2f}% {ap50:<11.2f}% {ap:<11.2f}%")

print("-" * 88)
avg_p = float(metrics.box.mp) * 100
avg_r = float(metrics.box.mr) * 100
avg_f1 = float(np.mean(f1_scores)) if f1_scores else 0.0
map50_all = float(metrics.box.map50) * 100
map_all = float(metrics.box.map) * 100
print(f"{'TỔNG HỢP (ALL)':<18} {avg_p:<11.2f}% {avg_r:<11.2f}% {avg_f1:<11.2f}% {map50_all:<11.2f}% {map_all:<11.2f}%")
print("=" * 88)

# 1. VẼ MA TRẬN NHẦM LẪN CHUẨN HÓA (ĐỌC ĐÚNG ULTRALYTICS: HÀNG LÀ PREDICTED, CỘT LÀ GROUND TRUTH)
try:
    cm = metrics.confusion_matrix.matrix # [pred, true]
    cm_norm = cm / (cm.sum(axis=0, keepdims=True) + 1e-6)
    labels = ["Pothole", "Crack", "Flooding", "Obstacle", "Background"]

    fig, ax = plt.subplots(figsize=(8, 6))
    cax = ax.matshow(cm_norm, cmap='Blues')
    plt.colorbar(cax)

    for i in range(len(labels)):
        for j in range(len(labels)):
            val = cm_norm[i, j]
            color = "white" if val > 0.5 else "black"
            ax.text(j, i, f"{val*100:.1f}%\n({int(cm[i, j])})", ha='center', va='center', color=color, fontsize=9)

    ax.set_xticks(range(len(labels)))
    ax.set_yticks(range(len(labels)))
    ax.set_xticklabels(labels, rotation=45, ha="left")
    ax.set_yticklabels(labels)
    plt.xlabel("Thực tế (Ground Truth) [Cột]", fontweight="bold")
    plt.ylabel("Dự đoán của mô hình (Predicted) [Hàng]", fontweight="bold")
    plt.title("MA TRẬN NHẦM LẪN CHUẨN HÓA (NORMALIZED RECALL CONFUSION MATRIX)", fontweight="bold", pad=20)
    plt.tight_layout()
    plt.show()

    gt_pothole_total = max(int(cm[:, 0].sum()), 1)
    gt_crack_total = max(int(cm[:, 1].sum()), 1)
    pothole_mis_as_crack = cm[1, 0] / gt_pothole_total
    crack_mis_as_pothole = cm[0, 1] / gt_crack_total
    # Tinh toan thuc te tren tap anh nen am ban (Negative background images)
    test_lbl_files = glob.glob(f"{base_dir}/labels/test/*.txt")
    bg_test_files = []
    for lf in test_lbl_files:
        try:
            with open(lf, 'r', encoding='utf-8') as f:
                if not f.read().strip():
                    bn = os.path.splitext(os.path.basename(lf))[0]
                    matched = glob.glob(f"{base_dir}/images/test/{bn}.*")
                    if matched:
                        bg_test_files.append(matched[0])
        except Exception:
            pass
    if bg_test_files:
        fp_bg_cnt = sum(1 for bf in bg_test_files if len(best_model.predict(bf, conf=0.30, verbose=False)[0].boxes) > 0)
        actual_fpr_bg = (fp_bg_cnt / len(bg_test_files)) * 100.0
    else:
        actual_fpr_bg = 0.0

    print("\n[PHAN TICH] PHÂN TÍCH NHẦM LẪN CHÉO KHOA HỌC (CROSS-CLASS ERROR ANALYSIS):")
    print(f"  • Pothole thật bị đoán nhầm thành Crack: {cm[1, 0]}/{gt_pothole_total} ({pothole_mis_as_crack*100:.2f}%)")
    print(f"  • Crack thật bị đoán nhầm thành Pothole: {cm[0, 1]}/{gt_crack_total} ({crack_mis_as_pothole*100:.2f}%)")
    print(f"  • Báo động giả trên nền đường sạch (FPR trên Background): {actual_fpr_bg:.2f}% (TNR = {100.0 - actual_fpr_bg:.2f}%)")
except Exception as e:
    print(f"⚠️ Lưu ý Confusion Matrix: {e}")

# ==============================================================================
# 2. BỘ TRỰC QUAN HÓA TOÀN DIỆN MỖI ĐỐI TƯỢNG TRÊN TẬP TEST
# ==============================================================================
print("\n" + "=" * 88)
print("[TRUC QUAN] ĐANG TRỰC QUAN HÓA CÁC TRƯỜNG HỢP KIỂM THỬ TRÊN TẬP TEST ĐỘC LẬP...")
print("=" * 88)

def get_images_by_filter(filter_fn, max_num=1):
    res_imgs = []
    lbl_files = sorted(glob.glob(f"{base_dir}/labels/test/*.txt"))
    for lbl in lbl_files:
        base_n = os.path.splitext(os.path.basename(lbl))[0]
        img_candidates = glob.glob(f"{base_dir}/images/test/{base_n}.*")
        if not img_candidates:
            continue
        img_p = img_candidates[0]
        try:
            with open(lbl, 'r', encoding='utf-8') as f:
                lines = [l.strip() for l in f if l.strip()]
            classes = [int(l.split()[0]) for l in lines]
            if filter_fn(base_n, classes, lines):
                res_imgs.append(img_p)
                if len(res_imgs) >= max_num:
                    break
        except Exception:
            pass
    return res_imgs

evaluation_categories = [
    {
        "title": "[O GA] LỚP 0 - POTHOLE: CÁC DẠNG Ổ GÀ TRÊN MẶT ĐƯỜNG",
        "subcases": [
            ("Ổ gà sâu / Đọng nước", lambda name, cls, lines: 0 in cls and 'pot_' in name),
            ("Ổ gà bong tróc nhựa đường", lambda name, cls, lines: 0 in cls and 'rdd_' in name),
            ("Vệt lún sâu vệt bánh xe (Rutting)", lambda name, cls, lines: 0 in cls and ('rut' in name or 'pot_vn' in name or 'pot_' in name))
        ]
    },
    {
        "title": "[VET NUT] LỚP 1 - ROAD_CRACK: CÁC DẠNG VẾT NỨT MẶT ĐƯỜNG",
        "subcases": [
            ("Vết nứt cận cảnh (CRACK500)", lambda name, cls, lines: 1 in cls and 'c500_' in name),
            ("Vết nứt mạng nhện mai rùa", lambda name, cls, lines: 1 in cls and 'rdd_' in name),
            ("Vết nứt rãnh dài theo làn đường", lambda name, cls, lines: 1 in cls and 'rdd_' in name)
        ]
    },
    {
        "title": "[NGAP NUOC] LỚP 2 - ROAD_FLOODING: NGẬP ÚNG LÒNG ĐƯỜNG",
        "subcases": [
            ("Ngập nước lòng đường diện rộng", lambda name, cls, lines: 2 in cls and 'flood_' in name),
            ("Phương tiện lội nước qua vùng ngập", lambda name, cls, lines: 2 in cls and 'flood_' in name)
        ]
    },
    {
        "title": "[VAT CAN] LỚP 3 - ROAD_OBSTACLE: VẬT CẢN (LỐP XE, CÂY ĐỔ, THÙNG XỐP, RÀO CHẮN)",
        "subcases": [
            ("Lốp xe rơi vãi / Giữ chỗ", lambda name, cls, lines: 3 in cls and ('tire' in name or 'lop_' in name)),
            ("Cây ngã đổ chắn đường", lambda name, cls, lines: 3 in cls and 'tree' in name),
            ("Thùng phuy phân làn / Cọc tiêu", lambda name, cls, lines: 3 in cls and ('hazard_' in name or 'barrel' in name)),
            ("Hố ga mất nắp nguy hiểm (Bẫy tử thần)", lambda name, cls, lines: 3 in cls and ('manhole' in name or 'open_manhole' in name or 'nap_cong' in name))
        ]
    },
    {
        "title": "[AN TOAN] MẪU NỀN ÂM TÍNH (NEGATIVE SAMPLES): TEST CHỐNG BÁO ĐỘNG GIẢ",
        "subcases": [
            ("Bóng cây & vệt nước rửa đường (Chuẩn 0 box)", lambda name, cls, lines: len(cls) == 0)
        ]
    }
]

for cat in evaluation_categories:
    sample_images = []
    sub_titles = []
    for sub_title, filter_fn in cat["subcases"]:
        imgs = get_images_by_filter(filter_fn, max_num=1)
        if imgs:
            sample_images.append(imgs[0])
            sub_titles.append(sub_title)

    if sample_images:
        n_imgs = len(sample_images)
        fig, axes = plt.subplots(1, n_imgs, figsize=(6 * n_imgs, 5.5))
        if n_imgs == 1: axes = [axes]
        fig.suptitle(cat["title"], fontsize=13, fontweight="bold", y=1.03)
        for idx, (img_p, sub_t) in enumerate(zip(sample_images, sub_titles)):
            res = best_model.predict(img_p, conf=0.25, imgsz=800)[0]
            res_im = cv2.cvtColor(res.plot(), cv2.COLOR_BGR2RGB)
            axes[idx].imshow(res_im)
            axes[idx].axis('off')
            num_boxes = len(res.boxes)
            fn_base = os.path.basename(img_p)
            if "NEGATIVE SAMPLES" in cat["title"]:
                status_txt = "[DAT CHUAN] 0 Box - KHÔNG BÁO ẢO!" if num_boxes == 0 else f"[CANH BAO] Báo {num_boxes} box ảo"
                axes[idx].set_title(f"{sub_t}\n[{fn_base}]\n{status_txt}", fontsize=10, fontweight="bold", color="green" if num_boxes == 0 else "red")
            else:
                confs = [f"{float(b.conf[0])*100:.0f}%" for b in res.boxes]
                conf_str = ", ".join(confs[:3]) + ("..." if len(confs) > 3 else "")
                axes[idx].set_title(f"{sub_t}\n[{fn_base}]\n[MUC TIEU] Phát hiện: {num_boxes} box ({conf_str})", fontsize=10, fontweight="bold")
        plt.tight_layout()
        plt.show()
        print("-" * 88)
```
